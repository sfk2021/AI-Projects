```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("/content/drive/MyDrive/Simple Linear Regression/data.csv")
df
```

```python
plt.xlabel("Height",color = "red")
plt.ylabel("Weight",color = "red")
plt.title("Height vs Weight",color = "red")
plt.scatter(df.Height,df.Weight)
```

```python
lr = linear_model.LinearRegression()
lr.fit(df[["Height"]],df.Weight)
```

```python
lr.predict([[2.00]])
```

```python
lr.score(df[['Height']],df.Weight)
lr.coef_
```

```python
lr.intercept_
```

```python

predicted = 61.27218654 * 2.00 + -39.06195591884392
print(predicted)
```

```python
plt.xlabel("Height",color = "red")
plt.ylabel("Weight",color = "red")
plt.title("Height vs Weight",color = "red")
plt.scatter(df.Height,df.Weight)
plt.plot(df.Height,lr.predict(df[["Height"]]),color = "blue")
```
